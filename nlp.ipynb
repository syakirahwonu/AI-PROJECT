```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')

text = "Ai in modern computing is powerful and interesting!"
tokens = word_tokenize(text)
print(tokens)
```

```python
#lemmatization

import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "studies", "leaves"]

for w in words:
  print(w + " -> " + lemmatizer.lemmatize(w))

```

```python
#example1 with word counter
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')

text = "Ai in modern computing is powerful and interesting!"
tokens = word_tokenize(text)
word_count = len(tokens)
print(word_count)
```

```python
#classactivity1
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')

text = "Although the main story concludes with Episode 8, fans will have more reasons to return to Kuala Makar. Astro has confirmed two bonus episodes that will be available exclusively on Astro and Sooka. The first, My Name is Ahyan Yasser, premieres on 3 July, offering a retelling of the events surrounding Ahyan's murder from a fresh perspective. This will be followed by The Story That Needed to Be Told on 10 July, giving viewers an opportunity to further explore the world and characters of Good Boys Go To Heaven."
tokens = word_tokenize(text)
word_count = len(tokens)
print(word_count)
```

```python
#classactivity2
from nltk.tokenize import sent_tokenize, word_tokenize

text = "Sementara itu, Pengerusi Ipoh Society for the Prevention of Cruelty to Animals (ISPCA) Ricky Soong berkata inisiatif berkenaan sedikit sebanyak membantu menggalakkan masyarakat mengambil 'haiwan angkat' dari pusat jagaan, sekali gus mengurangkan kesesakan pusat itu."
sentences = sent_tokenize(text)
tokens = word_tokenize(text)

word_count = len(tokens)
sentences_count = len(sentences)

print("Sentences count:", sentences_count)
print("Word count:", word_count)
```
