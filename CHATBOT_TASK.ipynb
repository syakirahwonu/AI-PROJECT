```python
print("Hi! How can I help you? (type 'exit' or 'quit' to stop)")
while True:
    user = input("> ").strip().lower()
    if user in ["exit", "quit"]:
        print("Goodbye!")
        break
    elif "penulis" in user or "author" in user:
        print("Mitch Albom")
    elif "book" in user or "buku" in user:
        print("He wrote Tuesdays with Morrie, widely regarded as the best-selling memoir of all time.")
    else:
        print("Sorry, I don't understand.")
```
